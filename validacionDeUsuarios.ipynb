```python

# Entrada de datos por parte del usuario.
nombre = input("Ingrese su Nombre: ")
apellido = input("Ingrese su apellido: ")
edad = input("Ingrese su edad: ")
correo_Electronico = input("Ingrese su Correo Electrónico: ")
```

```python
#ARREGLADO CREO SOLO FALTA ORGANIZAR ESTE CÓDIGO ESPAGUETTI
# Otra manera de comprobar la condicional if con len(nombre de la variable) == 0
# Operador lógico: (OR)

print("OR")
if len(edad) == 0:
  #-------------------------------------if anidado------------------------------------------------------------
  # Si nombre y apellido y correo electrónico son igual que 0: (TRUE)
  if len(nombre) == 0 or len(apellido) == 0 or len(correo_Electronico) == 0:
      print('ERROR!, No se admiten campos vacios.')
      print("PRIMER IF")
  else:
    print("ERROR! No se admiten campos vacíos")
    print("BLOQUE ELSE DEL PRIMER IF")
#-------------------------------------Bloque else-------------------------------------------------------------
else:
  print("BLOQUE ELSE DEL IF PRINCIPAL")
  #-------------------------------------if anidado------------------------------------------------------------
  if edad != '' and nombre != '' and apellido != '' and correo_Electronico != '':
    print("SEGUNDO IF")
    edad_int = int(edad)
    #-------------------------------------if anidado------------------------------------------------------------
    if edad_int >= 18:
      print("Es mayor de edad")
      print('Nombre: ' + nombre,
            'Apellido: ' + apellido,
            'Edad: ' + str(edad_int),
            'Correo Electrónico: ' + correo_Electronico, sep='\n')
    #Si no: (FALSE)
    else:
    #-------------------------------------if anidado------------------------------------------------------------
      if edad_int < 18:
        print("Es menor de edad, no tiene acceso al sitio.")
        print("TERCER IF SIN ELSE")
        #-------------------------------------if anidado------------------------------------------------------------
        #if len(nombre) == 0 or len(apellido) == 0 or len(correo_Electronico) == 0:
          #print('ERROR!, No se admiten campos vacios.')
          #print("CUARTO IF SIN ELSE")

    # Si no: (FALSE)
  else:
    print("BLOQUE ELSE DEL SEGUNDO IF")
    # Convierte el dato ingresado por el usuario del tipo str a int.
    edad_int = int(edad)
    print(type(edad_int))
    #-------------------------------------if anidado------------------------------------------------------------
    # Comprueba si la edad ingresada por el usuario es mayor o igual que 18. (TRUE)
    if edad_int >= 18:
      print("Es mayor de edad")
      if nombre != '' and apellido != '' and correo_Electronico != '':
        # Imprime lo siguiente:
        print('Nombre: ' + nombre,
              'Apellido: ' + apellido,
              # Imprime el string 'Edad:' y convierte el valor almacenado en la variable edad_int a str.
              'Edad: ' + str(edad_int),
              'Correo Electrónico: ' + correo_Electronico, sep='\n')
        print("EJECUCIÓN FINAL DEL QUINTO IF SIN ELSE")
      else:
        print("ERROR! No se admiten campos vacíos")
    # Si no. (FALSE)
    else:
      # Si la edad resulta ser menor a 18 imprime lo siguiente:
      print("Es menor de edad, no tiene acceso al sitio.")
```

```python
# Comprobación de entrada de datos y mayoría de edad.
# Operador lógico: (AND)

print("AND")
# Si nombre y apellido y edad y correo electrónico son igual que ''(vacío) = (TRUE)
if nombre == '' and apellido == '' and edad == '' and correo_Electronico == '':
  # Imprime lo siguiente:
  print("ERROR, No se admiten campos vacios")
  print("PRIMER IF PRINCIPAL")
#-------------------------------------Bloque else-------------------------------------------------------------
# Si no: (FALSE)
else:
  print("BLOQUE ELSE DEL IF PRINCIPAL")
  #-------------------------------------if anidado------------------------------------------------------------
  # Si edad es igual a vacío. (TRUE)
  if edad == '':
    # Imprime lo siguiente:
    print("ERROR, No se admiten campos vacios")
    print("SEGUNDO IF")
  #Si no. (FALSE)
  else:
    print("BLOQUE ELSE DEL SEGUNDO IF")
    # Convierte la variable edad, transformando su valor del tipo str a int, y luego la vuelve a almacenar en edad_int.
    edad_int = int(edad)
    #-------------------------------------if anidado------------------------------------------------------------
    # Compara si el valor de la variable edad_int (ahora del tipo int), es menor que 18. (TRUE)
    if edad_int < 18:
      # Imprime lo siguiente:
      print('Es menor de edad, no tiene acceso al sitio')
      print("TERCER IF")
      #-------------------------------------if anidado----------------------------------------------------------
      if nombre == '' or apellido == '' or correo_Electronico == '':
        print("ERROR! No se admiten campos vacíos")
        print("CUARTO IF SIN ELSE")
    # Si no. (FALSE)
    else:
      print("BLOQUE ELSE DEL TERCER IF")
      #-------------------------------------if anidado----------------------------------------------------------
      if nombre == '' or apellido == '' or correo_Electronico == '':
        # Imprime lo siguiente:
        print("ERROR, No se admiten campos vacios")
        print("QUINTO IF")
      else:
        print("BLOQUE ELSE DEL QUINTO IF")
        # Imprime lo siguiente:
        print("Es mayor de edad")
        # Imprime lo siguiente:
        print('Nombre: ' + nombre,
              'Apellido: ' + apellido,
              'Edad: ' + str(edad_int),
              'Correo Electrónico: ' + correo_Electronico, sep='\n')
        print("EJECUCIÓN FINAL EN EL BLOQUE ELSE DEL QUINTO IF")
```
