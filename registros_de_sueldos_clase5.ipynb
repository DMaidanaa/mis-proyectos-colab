## Ejercicio práctico

Necesitamos un software que ayude a registrar y calcular
información financiera básica para nuestros y nuestras clientes.
Tu tarea para esta semana es la siguiente:

1. Registrar los ingresos mensuales de un cliente durante 6 meses usando un bucle
while para solicitar el ingreso de cada mes. Validar que los ingresos sean números
positivos. Si se ingresa un valor negativo, mostrá un mensaje indicando que el
valor no es válido y volvé a pedir el dato.
2. Calcular el total acumulado durante los 6 meses y el promedio mensual. Mostrá
este resultado al final del programa.

```python
# Plazo de meses calculados.
meses = 6

# Variable contador que va iterando los meses.
mes = 1
print(f'Variable mes contiene el número: {mes}')

# Variable acumulador que va sumando los sueldos.
sueldos_mensuales_acumulados = 0
print(f'Sueldo acumulado es: {sueldos_mensuales_acumulados}')

# Mientras mes sea menor o igual que meses.
while mes <= meses:
  # Se solicita el dato del sueldo al usuario.
  sueldo = int(input(f"Ingresá tu sueldo mensual del mes {mes} (sin puntos, guiones o comas): "))

  # Si sueldo es mayor que 0 (o sea positivo)
  if sueldo > 0:
    # Asigna ese sueldo a la variable sueldos_mensuales_acumulados.
    sueldos_mensuales_acumulados += sueldo
    print(f'Sueldo acumulado del mes {mes} es de: {sueldos_mensuales_acumulados}')

    # Vuelve a pedir un nuevo ingreso de sueldo.
    # sueldo = int(input(f"Ingresa tu sueldo mensual del mes {mes} (sin puntos, guiones o comas): "))

    # Incrementá la variable contador mes en uno.
    mes += 1
  else:
    print("No se permite el número 0 ni números negativos")

print(f'El total de ingresos en {meses} meses es de: {sueldos_mensuales_acumulados}')
print(f'El promedio de tu sueldo en {meses} meses es de: {sueldos_mensuales_acumulados / meses:.2f}')
```
