```python
# Se pide los datos al User y al mismo tiempo se almacena en variables.
nombre = input("Ingrese su nombre: ")
apellido = input("Ingrese su apellido: ")
edad = int(input("Ingrese su edad: "))
correo_Electronico = input("Ingrese su correo electrónico: ")
```

```python
# Se muestran los resultados de los datos solicitados por consola o en pantalla
print('Nombre:', nombre)
print('Apellido:', apellido)
print('Edad:', edad)
print('Correo Electrónico:', correo_Electronico, "\n")

# Otra manera de mostrar los datos solicitados.
print(nombre, apellido, edad, correo_Electronico, sep='\n')
```

```python
# Usamos type() para identificar el tipo de cada variable

print("El tipo de 'nombre' es:", type(nombre))
print("El tipo de 'apellido' es:", type(apellido))
print("El tipo de 'edad' es:", type(edad))
print("El tipo de 'correo' es:", type(correo_Electronico))
```
