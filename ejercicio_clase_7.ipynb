*“Necesitamos implementar un registro ordenado de los nombres de las
empresas que trabajan con TalentoLab. Tu tarea hoy consistirá en crear
una lista, ordenada alfabéticamente, de todas ellas. Más tarde te daré

En **TalentoLab** necesitamos llevar un registro ordenado de los
nombres de los nuevos clientes y clientas que se van incorporando.
Queremos asegurarnos de que los datos ingresados sean válidos y
estén bien organizados. Tu tarea es escribir un programa en Python
que haga lo siguiente:

1. Solicite al usuario o usuaria los nombres de los clientes y clientas uno por uno y
valide que cada nombre no esté vacío. Si se deja el campo vacío, mostrale un
mensaje de advertencia y volvé a pedir el nombre.
2. Guarde cada nombre válido en una lista, asegurándote de agregarlo con el
método .append().
3. Permita que la persona finalice la carga de nombres escribiendo la palabra "fin".
4. Una vez finalizada la carga, ordene alfabéticamente los nombres en la lista y
muestre la lista ordenada de nombres utilizando un bucle for.

```python
# VER PORQUE EN CASOS DE UNA Z MAYÚSCULA IMPRIME PRIMERO ESO Y NO UNA A MINÚSCULA INGRESADA PARA QUE QUEDE BIEN ORDENADA.
# INVESTIGAR MÉTODO SORT KEY EN LA DOCUMENTACIÓN DE PYTHON.

nombre_clientes = input("Ingresa el nombre del cliente/a para agregar o 'fin' para terminar la lista: ").strip()

lista_clientes = []

while nombre_clientes.lower() != 'fin':

  if nombre_clientes == "":
    print("No se admiten registros vacios.")
    nombre_clientes = input("Vuelva a ingresar el nombre del cliente/a: ").strip()
  else:
    lista_clientes.append(nombre_clientes)
    nombre_clientes = input("Ingresa el nombre del cliente/a: ").strip()

print(f"Lista de clientes sin ordenar {lista_clientes}")

lista_clientes.sort()
print(f'Lista de clientes ordenada {lista_clientes}')

print("\nListado de nombres ordenados alfabeticamente:")
for nombre in lista_clientes:
  print(nombre)
```
