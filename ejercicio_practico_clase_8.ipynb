# **EJERCICIO PRÁCTICO DE LA CLASE 8:**

El equipo necesita una pequeña aplicación para registrar productos y sus precios en pesos. Esta vez, vas a necesitar una estructura eficiente para organizar y acceder rápidamente a los datos.

La mejor opción es usar diccionarios en Python, una herramienta que permite asociar claves con valores: vas a empezar a diseñar estructuras que se ajustan a los datos reales que maneja el equipo.

Nos han pedido desarrollar una aplicación que registre productos y sus precios utilizando diccionarios.
Necesitamos que tu programa cumpla con estas instrucciones:

● Crear un diccionario llamado productos donde las claves sean los nombres
de los productos y los valores sean sus precios.

● Permitir agregar productos y sus precios hasta que se decida finalizar.

● Mostrar el contenido del diccionario después de cada operación.

Recordá que los diccionarios te permiten acceder a los datos rápidamente usando claves y que podés modificar sus valores sin problemas. Aprovechá esto para hacer que tu programa sea eficiente y fácil de usar.

```python
# Creación de un diccionario de productos.
productos = {}

print(productos) # Salida: {}

while True:

  # Pedido del nombre del producto:
  nombre_producto = input("\nIngrese el nombre del producto [o presione enter para finalizar la carga]: ").lower().strip()

  if nombre_producto == "":
    break

  if nombre_producto in productos:
    print(f'{nombre_producto.capitalize()} ya existe en los registros.')
    continue

  # Si la condición resulta en un False ejecuta la siguiente variable que almacena lo que ingresó el usuario.
  precio_producto = int(input("Ingrese el precio del producto agregado sin comas, puntos ni guiones: "))

  # Se crea el par clave-valor almacenados en las variables "nombre_producto" y "precio_producto" y se agregan de a uno en el diccionario:
  productos[nombre_producto] = precio_producto

  # Estado actual del diccionario.
  print("\nEstado actual del diccionario Productos:")
  for nombre_producto,precio_producto in productos.items():
    print(f'Producto: {nombre_producto.capitalize()}, Precio: {precio_producto}')

# Estado final del diccionario.
print("\nEstado final del diccionario Productos:")
for nombre_producto,precio_producto in productos.items():
  print(f'Producto: {nombre_producto.capitalize()}, Precio: {precio_producto}')
```
