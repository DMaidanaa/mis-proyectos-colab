## Vas a desarrollar un programa en Python que cumpla con las siguientes características:

### Requerimientos:

1. Ingreso de datos de productos: El sistema debe permitir ingresar datos básicos de los productos: nombre, categoría, y precio (sin centavos). Estos datos deben almacenarse en una lista, donde cada producto sea representado/a como una sublista de tres elementos (nombre, categoría, y precio).

2. Visualización de productos registrados: El programa debe incluir una funcionalidad para mostrar en pantalla todos los productos ingresados. La información debe presentarse de manera ordenada y legible, con cada producto numerado.

3. Búsqueda de productos: El sistema debe permitir buscar productos por su nombre. Si encuentra coincidencias, debe mostrar la información completa de los productos que coincidan. Si no hay coincidencias, debe informar que no se encontraron resultados.

4. Eliminación de productos: El sistema debe permitir eliminar un producto de la lista, identificándolo por su posición (número) en la lista.

### Requisitos

* Usar listas para almacenar y gestionar los datos.

* Incorporar bucles while y for según corresponda.

* Validar entradas del usuario o usuaria, asegurándote de que no se ingresen datos vacíos o incorrectos.

* Utilizar condicionales para gestionar las opciones del menú y las validaciones necesarias.

* Presentar un menú que permita elegir entre las funcionalidades disponibles: agregar productos, visualizar productos, buscar productos y eliminar productos.

* El programa debe continuar funcionando hasta que se elija una opción para salir.

### Consejos

1. Usá lo aprendido sobre listas y bucles para gestionar los datos y recorrerlos.

2. Recordá validar las entradas utilizando condicionales.

3. Utilizá las herramientas vistas para organizar y presentar la información de manera clara.

```python
productos = [] # Lista vacía donde se va guardando el ingreso de los productos.

opcion_ingresada = 0 # Variable que sirve de contador.

while opcion_ingresada != 5:
  opcion_ingresada = int(input("Ingrese algún número de las siguientes opciones si desea: 1) Agregar | 2) Mostrar | 3) Buscar | 4) Eliminar | 5) Salir de la lista de Productos -->  "))
#-------------------------------------------------------------------------------------------------------
  match opcion_ingresada:
    case 1:
      # AGREGAR PRODUCTOS

      print("Elegiste agregar un producto")
      nombre_producto = input("Ingrese el nombre del producto: ").title().strip()

      while len(nombre_producto) == 0:
        print("ERROR! No se admiten registros vacíos")
        nombre_producto = input("Vuelva a ingresar el nombre del producto: ").title().strip()

      categoria_producto = input("Ingresa la categoría del producto: ").title().strip()

      while len(categoria_producto) == 0:
        print("ERROR! No se admiten registros vacíos")
        categoria_producto = input("Vuelva a ingresar la categoría del producto: ").title().strip()

      precio = int(input("Ingresa el precio del producto sin comas, puntos, guiones ni decimales: "))

      productos.append([nombre_producto, categoria_producto, precio])
    #----------------------------------------------------------------------------------------------------
    case 2:
      # MOSTRAR PRODUCTOS

      print("Elegiste mostrar productos")

      if not len(productos):
        print("ERROR!, No hay productos cargados. Ingrese la opción 1 para realizar la carga de los mismos.")

      else:

        contador = 1

        for producto in productos:
          print(f'Producto nro {contador}: {producto[0]}, Categoría: {producto[1]}, Precio: {producto[2]}')
          contador += 1
    #----------------------------------------------------------------------------------------------------
    case 3:
      # BUSCAR PRODUCTO

      print("Elegiste buscar un producto")
      producto_buscado = input("Ingrese el nombre del producto buscado: ").title().strip()

      producto_encontrado = False

      for producto in productos:
        if producto[0] == producto_buscado:
          print(f'Aquí está tu producto: {producto[0]}, Categoría: {producto[1]}, Precio: {producto[2]}')
          producto_encontrado = True
          break

      if not producto_encontrado:
        print("Producto no disponible")
    #----------------------------------------------------------------------------------------------------
    case 4:
      # ELIMINAR PRODUCTO

      print("Elegiste eliminar un producto")
      producto_a_eliminar = int(input(f"Ingresa el número del producto a eliminar de los {len(productos)} registrados: "))

      producto_eliminado = False

      contador = 1

      for producto in productos:
        if producto_a_eliminar == contador:
          productos.remove(producto)
          producto_eliminado = True
          print(f'El producto: {producto[0]}, registrado con el número: {contador} fue eliminado de la lista.')
          break

        contador += 1

      if not producto_eliminado:
        print(f'El producto registrado con el nro: {producto_a_eliminar} no se encuentra en la lista')
    #----------------------------------------------------------------------------------------------------
    case 5:
      print("Elegiste la opción salir")
      break
    #----------------------------------------------------------------------------------------------------
    case _:
      print("Opción inválida, ingrese solamente las opciones indicadas en el menú")
```
